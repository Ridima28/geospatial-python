```python
api_url = "https://cmr.earthdata.nasa.gov/stac/LPCLOUD"

from pystac_client import Client

client = Client.open(api_url)

collection = client.get_collections()

for i in collection:
    print(i.id)
```

```python
from shapely.geometry import Point
collection_HLSL30_v2 = "HLSL30.v2.0" 

point = Point(-73.97, 40.78)
```

```python
search = client.search(collections = [collection_HLSL30_v2], 
                    datetime = "2021-02-01/2021-03-31",
                    intersects = point)
```

```python
items = search.item_collection()
assets = items[-1].assets
print(assets.keys())
```
